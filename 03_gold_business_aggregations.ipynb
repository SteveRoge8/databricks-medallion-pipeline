```python
%sql
USE CATALOG workspace;

-- 1. Create a Gold level aggregated table using a CTAS (Create Table As Select) statement
CREATE OR REPLACE TABLE gold.daily_sales_revenue
COMMENT 'Daily aggregated revenue and order volume for executive BI dashboards'
AS
SELECT 
    CAST(o.order_purchase_timestamp AS DATE) AS sale_date,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(i.price), 2) AS total_product_revenue,
    ROUND(SUM(i.freight_value), 2) AS total_freight_revenue,
    ROUND(SUM(i.price) + SUM(i.freight_value), 2) AS total_revenue
FROM silver.orders o
JOIN silver.order_items i 
    ON o.order_id = i.order_id
WHERE o.order_purchase_timestamp IS NOT NULL
GROUP BY CAST(o.order_purchase_timestamp AS DATE)
ORDER BY sale_date DESC;

-- 2. Query the resulting Gold table to verify
SELECT * FROM gold.daily_sales_revenue LIMIT 10;
```

```python
%sql
-- Fetch the revenue trend over time, ignoring the limit so we see the full curve
SELECT 
    sale_date, 
    total_revenue 
FROM gold.daily_sales_revenue 
ORDER BY sale_date;
```
